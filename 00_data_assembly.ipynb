

Notebook 0: Master Dataset Preparation

Purpose:
To assemble raw UIDAI demographic, enrolment, and biometric datasets
into consolidated master files for downstream analysis.

Outputs:
- demographic_master.csv
- enrolment_master.csv
- biometric_master.csv

Scope:
This notebook performs data loading and concatenation only.
No analysis, aggregation, or inference is conducted here.


```python
import os

print("\nworking directory set to:")
print(os.getcwd())

expected_folders = ['demographic', 'enrolment', 'biometric']

print("\nInput folder verification::")
for folder in expected_folders:
    if os.path.exists(folder):
        files = os.listdir(folder)
        print(f"'{folder}'  {len(files)} files detected")
    else:
        print(f" '{folder}' NOT FOUND")

```

```python
import os
import pandas as pd

demo_path = "demographic"
demo_files = os.listdir(demo_path)

print("\nDemographic source files identified:")
for f in demo_files:
    print(f"- {f}")

demo_dfs = []

for f in demo_files:
    df = pd.read_csv(os.path.join(demo_path, f))
    demo_dfs.append(df)
    print("File loaded:")
    print("Rows:", df.shape[0])
    print("Columns:", df.columns.tolist())

print("\nTotal demographic files loaded:", len(demo_dfs))
```

```python
import pandas as pd
demographic_master = pd.concat(demo_dfs, ignore_index=True)

print("Demographic master dataset  assembled")
print("Total rows:", demographic_master.shape[0])
print("Total columns:", demographic_master.shape[1])

print("\nSample records (first 5 rows):")
print(demographic_master.head())

print("\nOutput written to file: demographic_master.csv")
```

```python
import os
import pandas as pd

enrol_path = "enrolment"
enrol_files = os.listdir(enrol_path)

print("Enrolment files found:")
for f in enrol_files:
    print(" -", f)

enrol_dfs = []

for file in enrol_files:
    df = pd.read_csv(os.path.join(enrol_path, file))
    enrol_dfs.append(df)
    print(f"File loaded: {file}")
    print("Rows:", df.shape[0])
    print("Columns:", df.columns.tolist())


print(f"\nTotal enrolment files loaded: {len(enrol_dfs)}")
```

```python
enrolment_master = pd.concat(enrol_dfs, ignore_index=True)

print("Enrolment master dataset assembled.")
print("Total rows:", enrolment_master.shape[0])
print("Total columns:", enrolment_master.shape[1])

print("\nSample records (first 5 rows):")
print(enrolment_master.head())

print("\nOutput written to file: enrolment_master.csv")
```

```python
biometric_master = pd.concat(bio_dfs, ignore_index=True)

print("Biometric master dataset assembled.")
print("Total rows:", biometric_master.shape[0])
print("Total columns:", biometric_master.shape[1])

print("\nSample records (first 5 rows):")
print(biometric_master.head())

print("\nOutput written to file: biometric_master.csv")

```
